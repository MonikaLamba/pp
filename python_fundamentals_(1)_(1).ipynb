## 1. Identifiers

An **identifier** is simply a **name** you give to a value so you can use it later.

Let's make a name called `name` and store the text `Rahul` in it, then show it:

```python
name = "Rahul"
print(name)
```

You can make as many names as you like. Each one holds one value:

```python
age = 21
city = "Delhi"
print(age)
print(city)
```

**Rules for a name:**
- Use letters, numbers, and the underscore `_`
- It **cannot start with a number**
- No spaces and no symbols like `@ # $ -`

These names follow the rules, so they work fine:

```python
_count = 10
total_marks = 495
student2 = "Aisha"
print(_count, total_marks, student2)
```

Not sure if a name is allowed? Ask Python with `.isidentifier()`.
`True` means allowed, `False` means not allowed:

```python
print("marks".isidentifier())     # good name
print("2marks".isidentifier())    # starts with a number -> not allowed
```

Names are **case-sensitive**, so `age` and `Age` are two different names:

```python
age = 21
Age = 22
print(age)
print(Age)
```

## 2. Keywords

**Keywords** are special words that already have a job in Python (like `if`, `for`,
`class`). You are **not allowed** to use them as your own names.

Python keeps the list of keywords ready for us. Let's see how many there are:

```python
import keyword
print(len(keyword.kwlist))
```

Here is the full list of keywords:

```python
print(keyword.kwlist)
```

We can ask Python whether a word is a keyword:

```python
print(keyword.iskeyword("if"))       # True  -> it's a keyword
print(keyword.iskeyword("marks"))    # False -> safe to use as a name
```

## 3. Escape Sequences

An **escape sequence** is a backslash `\` plus a letter. It lets you add special
things inside text, like a new line or a tab.

`\n` moves the text to a **new line**:

```python
print("Line1\nLine2")
```

`\t` adds a **tab** (a big space), handy for lining things up:

```python
print("Name:\tRahul")
```

To print a real backslash, write it **twice** `\\`:

```python
print("C:\\Users\\Rahul")
```

Need a quote inside quotes? Put a backslash before it:

```python
print('It\'s Python')
print("She said \"Hi\"")
```

Add the letter `r` before a string to keep every backslash exactly as typed.
This is great for file paths:

```python
print(r"C:\new\table.txt")
```

## 4. Data Types

A **data type** is the *kind* of a value. Use `type()` to check any value.

Whole numbers are called `int` (integer):

```python
x = 100
print(type(x))
```

Numbers with a decimal point are `float`:

```python
y = 3.14
print(type(y))
```

Text is called `str` (string). Always wrap text in quotes:

```python
word = "Python"
print(type(word))
```

`True` and `False` values are called `bool`. A comparison gives a bool:

```python
print(type(True))
print(10 > 5)
```

A `list` holds many values in order, inside square brackets `[ ]`:

```python
nums = [1, 2, 3]
print(nums)
print(type(nums))
```

A `dict` stores pairs of **key → value**, inside curly braces `{ }`:

```python
student = {"name": "Rahul", "age": 21}
print(student["name"])
print(student["age"])
```

You can change one type into another. Here we turn the text `"25"` into a number
so we can do math with it:

```python
print(int("25"))
print(int("25") + 5)
```

## 5. Variables

A **variable** is a name that points to a value. In Python you just write
`name = value` — you never have to say what type it is.

Create a variable and print it:

```python
marks = 95
print(marks)
```

You can change what a variable holds at any time:

```python
marks = 95
print(marks)

marks = 100
print(marks)
```

A variable can even change to a completely different type:

```python
data = 42
print(data)

data = "hello"
print(data)
```

Variables are most useful in calculations. Try changing the numbers:

```python
price = 50
quantity = 3
print(price * quantity)
```

## 6. Assignment

**Assignment** means putting a value into a variable using the `=` sign.

*(Note: `=` means "put into". Later you'll see `==` which means "is equal to".)*

The simplest form — one value into one variable:

```python
a = 5
print(a)
```

Give the **same** value to several names at once:

```python
x = y = z = 0
print(x, y, z)
```

Give **different** values in a single line:

```python
name, age = "Rahul", 21
print(name, age)
```

A neat trick — swap two values in one line, no helper needed:

```python
p, q = 1, 2
p, q = q, p
print(p, q)
```

`+=` is a short way to add to a variable. `n += 5` means `n = n + 5`:

```python
n = 10
n += 5
print(n)
```

## 7. Operators and Operands

In `5 + 3`, the `+` is the **operator** (the action) and `5` and `3` are the
**operands** (the values it works on).

Let's set two numbers to work with, then **add** them with `+`:

```python
a = 12
b = 5
print(a + b)
```

Subtract `-`, multiply `*`, and divide `/`:

```python
print(a - b)
print(a * b)
print(a / b)
```

`//` gives the whole-number part of a division, `%` gives the remainder:

```python
print(a // b)   # 12 divided by 5 is 2 (and a bit)
print(a % b)    # the leftover is 2
```

`**` means power. `a ** b` is `a` multiplied by itself `b` times:

```python
print(a ** b)
```

**Comparison** operators ask a question and answer `True` or `False`:

```python
print(a > b)     # is a bigger than b?
print(a == b)    # is a equal to b?
```

**Logical** operators join conditions: `and`, `or`, `not`:

```python
print(True and False)
print(True or False)
print(not True)
```

`in` checks whether a value is inside a list:

```python
fruits = ["apple", "banana", "cherry"]
print("apple" in fruits)
print("mango" in fruits)
```
